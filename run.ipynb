```python

import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay

# Load predictions generated by the trained SVM model
results = pd.read_csv("outputs/svm_predictions.csv")

emotion_labels = [
    "angry", "calm", "disgust", "fearful",
    "happy", "neutral", "sad", "surprised"
]

# Calculate confusion matrix
cm = confusion_matrix(
    results["actual_emotion"],
    results["predicted_emotion"],
    labels=emotion_labels
)

# Draw the graph
fig, ax = plt.subplots(figsize=(10, 8))

ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=emotion_labels
).plot(
    ax=ax,
    cmap="Blues",
    values_format="d",
    xticks_rotation=45
)

plt.title("SVM Confusion Matrix - RAVDESS")
plt.xlabel("Predicted Emotion")
plt.ylabel("Actual Emotion")
plt.tight_layout()

# Save the image
Path("outputs").mkdir(exist_ok=True)
plt.savefig(
    "outputs/confusion_matrix.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close(fig)

print("Confusion matrix saved successfully!")

```
